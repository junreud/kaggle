## 1단계: 데이터 로딩 및 초기 설정

실무에서는 먼저 작업 환경을 체계적으로 설정합니다.

```python
# 필수 라이브러리 import
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
import warnings
warnings.filterwarnings('ignore')

# 시각화 설정
plt.style.use('default')
sns.set_palette("husl")
plt.rcParams['figure.figsize'] = (10, 6)
plt.rcParams['font.size'] = 12

# pandas 표시 옵션 설정
pd.set_option('display.max_columns', None)
pd.set_option('display.max_rows', 100)

print("✅ 라이브러리 로딩 완료!")
print("📁 다음 단계: 데이터 로딩")
```

```python
# 데이터 경로 설정 (실무에서는 이런 식으로 경로를 체계화합니다)
DATA_PATH = '../data/raw/'
PROCESSED_PATH = '../data/processed/'

# TODO: 여기에 데이터를 다운로드하거나 로딩하는 코드를 작성해보세요
# 지금은 아직 데이터가 없으니 다음 셀에서 데이터 다운로드 방법을 안내하겠습니다

print(f"🗂️  원본 데이터 경로: {DATA_PATH}")
print(f"🔄 처리된 데이터 경로: {PROCESSED_PATH}")
print("⚠️  아직 데이터가 없습니다. 다음 셀에서 데이터를 준비해보겠습니다.")
```

```python
# 데이터 파일 확인 (실무에서는 항상 먼저 확인합니다)
import os
print("📂 data/raw 폴더의 파일들:")
for file in os.listdir(DATA_PATH):
    if file.endswith('.csv'):
        file_path = os.path.join(DATA_PATH, file)
        file_size = os.path.getsize(file_path)
        print(f"  - {file}: {file_size:,} bytes")

print("\n✅ 모든 필요한 파일이 준비되었습니다!")
```

```python
# 데이터 로딩 (실무에서는 각 파일의 역할을 명확히 합니다)
print("🔄 데이터 로딩 중...")

# 훈련 데이터 (목표변수 포함)
train_df = pd.read_csv(DATA_PATH + 'train.csv')
print(f"📊 Train 데이터: {train_df.shape}")

# 테스트 데이터 (목표변수 없음)
test_df = pd.read_csv(DATA_PATH + 'test.csv')
print(f"🔍 Test 데이터: {test_df.shape}")

# 제출 양식 (예측해야 할 승객 ID와 생존 여부)
submission_df = pd.read_csv(DATA_PATH + 'gender_submission.csv')
print(f"📝 Submission 양식: {submission_df.shape}")

print(f"\n✅ 데이터 로딩 완료!")
print(f"💡 총 {len(train_df) + len(test_df)}명의 승객 데이터 (훈련: {len(train_df)}, 테스트: {len(test_df)})")
```

## 2단계: 첫 번째 데이터 탐색 (First Look)

실무에서는 데이터를 로딩한 후 바로 기본적인 탐색을 시작합니다. 
먼저 "이 데이터가 어떻게 생겼는지" 파악하는 것이 중요해요.

```python
# 훈련 데이터의 첫 번째 관찰 (실무에서 가장 먼저 하는 일)
print("🔍 훈련 데이터 첫 5행:")
print(train_df.head())

print("\n" + "="*60)
print("📊 기본 정보:")
print(f"- 행(승객 수): {train_df.shape[0]:,}명")  # .shape: (행 수, 열 수) 튜플로 반환
print(f"- 열(특성 수): {train_df.shape[1]}개")
print(f"- 목표변수(Survived): 0=사망, 1=생존")
    
print(f"\n📈 생존율 요약:")
survival_rate = train_df['Survived'].mean()
print(f"- 전체 생존율: {survival_rate:.1%} ({train_df['Survived'].sum()}명/{len(train_df)}명)")
print(f"- 사망: {(1-survival_rate):.1%}, 생존: {survival_rate:.1%}")
```

```python
# 데이터 타입과 구조 확인 (실무에서 반드시 체크하는 부분)
print("🔧 데이터 타입과 누락값 정보:")
print(train_df.info())

print("\n" + "="*60)
print("❗ 누락값(Missing Values) 요약:")
missing_data = train_df.isnull().sum()
missing_percent = (missing_data / len(train_df)) * 100

missing_summary = pd.DataFrame({
    '누락 개수': missing_data,
    '누락 비율(%)': missing_percent.round(1)
})
missing_summary = missing_summary[missing_summary['누락 개수'] > 0].sort_values('누락 개수', ascending=False)

if len(missing_summary) > 0:
    print(missing_summary)
    print("\n💡 실무 팁: 누락값이 많은 컬럼은 특별한 처리가 필요합니다!")
else:
    print("✅ 누락값이 없습니다!")
```

## 3단계: 핵심 특성별 생존율 분석

실무에서는 기본 탐색 후, **생존에 영향을 미치는 주요 요인들**을 하나씩 분석합니다.
각 특성이 생존에 어떤 영향을 미치는지 파악하는 것이 모델링의 핵심입니다.

### 🚺🚹 성별별 생존율 분석
타이타닉에서는 "여성과 어린이 먼저" 원칙이 있었습니다. 실제 데이터에서 확인해보겠습니다.

```python
# 성별별 생존율 분석
print("🚺🚹 성별별 생존율 분석")
print("=" * 50)

# 성별별 생존율 계산
gender_survival = train_df.groupby('Sex')['Survived'].agg(['count', 'sum', 'mean']).round(3)
gender_survival.columns = ['총 인원', '생존자 수', '생존율']
gender_survival['생존율(%)'] = (gender_survival['생존율'] * 100).round(1)

print(gender_survival)

# 요약
print(f"\n💡 주요 발견:")
female_survival = gender_survival.loc['female', '생존율(%)']
male_survival = gender_survival.loc['male', '생존율(%)']
print(f"- 여성 생존율: {female_survival}%")
print(f"- 남성 생존율: {male_survival}%")
print(f"- 차이: {female_survival - male_survival:.1f}%포인트")

if female_survival > male_survival:
    print("✅ 여성의 생존율이 훨씬 높습니다! (여성/어린이 우선 원칙 확인)")
else:
    print("⚠️  예상과 다른 결과입니다.")
```

### 🎫 객실등급별 생존율 분석  
사회경제적 지위(객실등급)가 생존에 미치는 영향을 분석해보겠습니다.

```python
# 객실등급별 생존율 분석
print("🎫 객실등급별 생존율 분석")
print("=" * 50)

# 객실등급별 생존율 계산
class_survival = train_df.groupby('Pclass')['Survived'].agg(['count', 'sum', 'mean']).round(3)
class_survival.columns = ['총 인원', '생존자 수', '생존율']
class_survival['생존율(%)'] = (class_survival['생존율'] * 100).round(1)

# 등급명 추가 (실무에서는 이해하기 쉽게 만듭니다)
class_names = {1: '1등실(상류층)', 2: '2등실(중산층)', 3: '3등실(서민층)'}
class_survival['등급'] = [class_names[i] for i in class_survival.index]
class_survival = class_survival[['등급', '총 인원', '생존자 수', '생존율(%)']]

print(class_survival)

# 평균 요금도 함께 확인
print(f"\n💰 등급별 평균 요금:")
avg_fare = train_df.groupby('Pclass')['Fare'].mean().round(2)
for pclass, fare in avg_fare.items():
    print(f"- {class_names[pclass]}: ${fare:.2f}")

print(f"\n💡 주요 발견:")
print(f"- 1등실 → 3등실로 갈수록 생존율 감소")
print(f"- 사회경제적 지위가 생존에 큰 영향을 미침")
```

### 👶👴 나이대별 생존율 분석
"여성과 어린이 먼저" 원칙에서 어린이 부분을 확인해보겠습니다.

```python
# 나이대별 생존율 분석 (누락값 있음 주의)
print("👶👴 나이대별 생존율 분석")
print("=" * 50)

# 나이 데이터가 있는 사람들만 분석
age_data = train_df[train_df['Age'].notna()].copy()

# 나이대 구분 (실무에서 자주 사용하는 방식)
def categorize_age(age):
    if age < 18:
        return '어린이 (<18세)'
    elif age < 30:
        return '청년 (18-29세)'
    elif age < 50:
        return '중년 (30-49세)'
    else:
        return '노년 (50세+)'

age_data['age_group'] = age_data['Age'].apply(categorize_age)

# 나이대별 생존율 계산
age_survival = age_data.groupby('age_group')['Survived'].agg(['count', 'sum', 'mean']).round(3)
age_survival.columns = ['총 인원', '생존자 수', '생존율']
age_survival['생존율(%)'] = (age_survival['생존율'] * 100).round(1)

# 순서대로 정렬
age_order = ['어린이 (<18세)', '청년 (18-29세)', '중년 (30-49세)', '노년 (50세+)']
age_survival = age_survival.reindex(age_order)

print(age_survival)

print(f"\n💡 주요 발견:")
child_survival = age_survival.loc['어린이 (<18세)', '생존율(%)']
print(f"- 어린이 생존율: {child_survival}%")
print(f"- 어린이 우선 원칙이 실제로 적용되었는지 확인")
print(f"⚠️  주의: 나이 데이터 누락이 {train_df['Age'].isnull().sum()}명 있음")
```

### 🎯 핵심 인사이트 종합

지금까지의 분석 결과를 종합해서 **어떤 특성이 생존에 가장 중요한지** 파악해보겠습니다.

```python
# 핵심 특성들의 조합 분석 (실무에서 중요한 단계)
print("🎯 성별 + 객실등급 조합 분석")
print("=" * 60)

# 성별과 객실등급을 조합한 생존율
combo_survival = train_df.groupby(['Sex', 'Pclass'])['Survived'].agg(['count', 'mean']).round(3)
combo_survival.columns = ['인원수', '생존율']
combo_survival['생존율(%)'] = (combo_survival['생존율'] * 100).round(1)

print(combo_survival)

print(f"\n💡 핵심 발견:")
print(f"✅ 가장 높은 생존율 그룹 확인")
print(f"❌ 가장 낮은 생존율 그룹 확인")
print(f"🔄 성별과 계층의 상호작용 효과")

# 간단한 베이스라인 모델을 위한 규칙 생성
print(f"\n🤖 간단한 예측 규칙 (베이스라인):")
print(f"- 여성 → 생존 예측")
print(f"- 남성 + 1등실 → 생존 예측") 
print(f"- 나머지 → 사망 예측")
print(f"이 규칙의 정확도를 다음 셀에서 계산해보겠습니다!")
```

## 4단계: 첫 번째 베이스라인 모델

실무에서는 복잡한 모델 전에 **간단하지만 합리적인 규칙**부터 시작합니다.
이것이 나중에 개발할 ML 모델의 성능 비교 기준이 됩니다.

```python
# 규칙 기반 베이스라인 모델 생성
print("🤖 규칙 기반 베이스라인 모델")
print("=" * 50)

def baseline_prediction(row):
    """
    분석 결과를 바탕으로 한 간단한 예측 규칙:
    1. 여성이면 생존
    2. 남성이지만 1등실이면 생존  
    3. 나머지는 사망
    """
    if row['Sex'] == 'female':
        return 1  # 생존
    elif row['Sex'] == 'male' and row['Pclass'] == 1:
        return 1  # 생존
    else:
        return 0  # 사망

# 예측 수행
train_df['baseline_pred'] = train_df.apply(baseline_prediction, axis=1)

# 정확도 계산
accuracy = (train_df['Survived'] == train_df['baseline_pred']).mean()

print(f"📊 베이스라인 모델 성능:")
print(f"- 정확도: {accuracy:.3f} ({accuracy*100:.1f}%)")

# 혼동 행렬 (간단 버전)
from sklearn.metrics import confusion_matrix, classification_report

print(f"\n📈 상세 성능:")
print(classification_report(train_df['Survived'], train_df['baseline_pred'], 
                          target_names=['사망', '생존']))

print(f"\n💡 베이스라인 해석:")
print(f"- 이 간단한 규칙만으로도 {accuracy*100:.1f}% 정확도!")
print(f"- 복잡한 ML 모델은 이보다 더 좋아야 함")
print(f"- 실무에서는 이런 베이스라인이 매우 중요")
```
